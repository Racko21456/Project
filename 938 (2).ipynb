```python
import os
import glob
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import joblib
import warnings

warnings.filterwarnings("ignore")

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import LabelEncoder, StandardScaler, OneHotEncoder
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score
from sklearn.metrics import confusion_matrix, classification_report, roc_auc_score, roc_curve
from sklearn.ensemble import RandomForestClassifier

from xgboost import XGBClassifier
```

```python
DATASET_NAME = "unsw"       # options: "unsw", "cicids2017", "cicids2018"
TARGET_COLUMN = "label"     # UNSW = label, CICIDS = Label
MAX_ROWS = 80000            # reduce to 30000 if Colab crashes
TEST_SIZE = 0.25
RANDOM_STATE = 42

os.makedirs("/content/outputs", exist_ok=True)
```

```python
def clean_columns(df):
    df = df.copy()
    df.columns = [str(col).strip().replace(" ", "_") for col in df.columns]
    return df


def load_unsw(max_rows=80000):
    train_path = "/content/datasets/unsw_nb15/UNSW_NB15_training-set.csv"
    test_path = "/content/datasets/unsw_nb15/UNSW_NB15_testing-set.csv"

    train_df = pd.read_csv(train_path)
    test_df = pd.read_csv(test_path)

    df = pd.concat([train_df, test_df], ignore_index=True)
    df = clean_columns(df)

    if max_rows is not None and len(df) > max_rows:
        df = df.sample(n=max_rows, random_state=42).reset_index(drop=True)

    return df


def load_cicids(folder_path, max_rows=80000):
    csv_files = glob.glob(folder_path + "/**/*.csv", recursive=True)

    print("CSV files found:", len(csv_files))
    for file in csv_files[:10]:
        print(file)

    frames = []

    rows_per_file = None
    if max_rows is not None and len(csv_files) > 0:
        rows_per_file = max(1000, max_rows // len(csv_files))

    for file in csv_files:
        try:
            temp = pd.read_csv(file, nrows=rows_per_file, low_memory=False)
            temp = clean_columns(temp)
            temp["source_file"] = os.path.basename(file)
            frames.append(temp)
            print("Loaded:", file, temp.shape)
        except Exception as e:
            print("Skipped:", file, e)

    df = pd.concat(frames, ignore_index=True)

    if max_rows is not None and len(df) > max_rows:
        df = df.sample(n=max_rows, random_state=42).reset_index(drop=True)

    return df


if DATASET_NAME == "unsw":
    df = load_unsw(MAX_ROWS)
    TARGET_COLUMN = "label"

elif DATASET_NAME == "cicids2017":
    df = load_cicids("/content/datasets/cicids2017", MAX_ROWS)
    TARGET_COLUMN = "Label"

elif DATASET_NAME == "cicids2018":
    df = load_cicids("/content/datasets/cicids2018", MAX_ROWS)
    TARGET_COLUMN = "Label"

else:
    raise ValueError("Invalid dataset name.")

print("Dataset loaded successfully.")
print("Shape:", df.shape)
print(df.head())
print(df.columns.tolist())
```

```python
df = clean_columns(df)
TARGET_COLUMN = TARGET_COLUMN.strip().replace(" ", "_")

print("Target column:", TARGET_COLUMN)

if TARGET_COLUMN not in df.columns:
    print("Available columns:")
    print(df.columns.tolist())
    raise ValueError("Target column not found. Check the exact label column name.")

df = df.replace([np.inf, -np.inf], np.nan)
df = df.dropna(subset=[TARGET_COLUMN])

print("Class distribution:")
print(df[TARGET_COLUMN].value_counts())
```

```python
drop_columns = [
    TARGET_COLUMN,
    "id",
    "ID",
    "Flow_ID",
    "Timestamp",
    "timestamp",
    "Src_IP",
    "Dst_IP",
    "Source_IP",
    "Destination_IP",
    "source_file"
]

X = df.drop(columns=[col for col in drop_columns if col in df.columns], errors="ignore")
y_raw = df[TARGET_COLUMN].astype(str).str.strip()

label_encoder = LabelEncoder()
y = label_encoder.fit_transform(y_raw)

print("Features shape:", X.shape)
print("Target classes:", label_encoder.classes_)
print("Encoded classes:", np.unique(y))
```

```python
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=TEST_SIZE,
    random_state=RANDOM_STATE,
    stratify=y
)

print("Training data:", X_train.shape)
print("Testing data:", X_test.shape)
```

```python
numeric_features = X_train.select_dtypes(include=["int64", "float64", "int32", "float32", "bool"]).columns.tolist()
categorical_features = [col for col in X_train.columns if col not in numeric_features]

print("Numeric features:", len(numeric_features))
print("Categorical features:", len(categorical_features))

try:
    onehot = OneHotEncoder(handle_unknown="ignore", sparse_output=False)
except TypeError:
    onehot = OneHotEncoder(handle_unknown="ignore", sparse=False)

numeric_transformer = Pipeline(steps=[
    ("imputer", SimpleImputer(strategy="median")),
    ("scaler", StandardScaler())
])

categorical_transformer = Pipeline(steps=[
    ("imputer", SimpleImputer(strategy="most_frequent")),
    ("onehot", onehot)
])

preprocessor = ColumnTransformer(
    transformers=[
        ("num", numeric_transformer, numeric_features),
        ("cat", categorical_transformer, categorical_features)
    ]
)
```

```python
def evaluate_model(model, model_name, X_test, y_test):
    y_pred = model.predict(X_test)

    if hasattr(model, "predict_proba"):
        y_proba = model.predict_proba(X_test)
    else:
        y_proba = None

    average_type = "binary" if len(label_encoder.classes_) == 2 else "weighted"

    accuracy = accuracy_score(y_test, y_pred)
    precision = precision_score(y_test, y_pred, average=average_type, zero_division=0)
    recall = recall_score(y_test, y_pred, average=average_type, zero_division=0)
    f1 = f1_score(y_test, y_pred, average=average_type, zero_division=0)

    try:
        if y_proba is not None and len(label_encoder.classes_) == 2:
            roc_auc = roc_auc_score(y_test, y_proba[:, 1])
        elif y_proba is not None:
            roc_auc = roc_auc_score(y_test, y_proba, multi_class="ovr", average="weighted")
        else:
            roc_auc = None
    except:
        roc_auc = None

    cm = confusion_matrix(y_test, y_pred)

    print("\n==============================")
    print(model_name)
    print("==============================")
    print("Accuracy:", accuracy)
    print("Precision:", precision)
    print("Recall:", recall)
    print("F1 Score:", f1)
    print("ROC AUC:", roc_auc)
    print("\nClassification Report:")
    print(classification_report(y_test, y_pred, target_names=label_encoder.classes_, zero_division=0))

    plt.figure(figsize=(8, 6))
    plt.imshow(cm)
    plt.title(f"Confusion Matrix - {model_name}")
    plt.xlabel("Predicted")
    plt.ylabel("Actual")
    plt.colorbar()

    ticks = np.arange(len(label_encoder.classes_))
    plt.xticks(ticks, label_encoder.classes_, rotation=45, ha="right")
    plt.yticks(ticks, label_encoder.classes_)

    for i in range(cm.shape[0]):
        for j in range(cm.shape[1]):
            plt.text(j, i, str(cm[i, j]), ha="center", va="center")

    plt.tight_layout()
    plt.savefig(f"/content/outputs/confusion_matrix_{model_name}.png", dpi=300)
    plt.show()

    with open(f"/content/outputs/classification_report_{model_name}.txt", "w") as f:
        f.write(classification_report(y_test, y_pred, target_names=label_encoder.classes_, zero_division=0))

    return {
        "model": model_name,
        "accuracy": accuracy,
        "precision": precision,
        "recall": recall,
        "f1_score": f1,
        "roc_auc": roc_auc
    }
```

```python
rf_model = Pipeline(steps=[
    ("preprocessor", preprocessor),
    ("classifier", RandomForestClassifier(
        n_estimators=150,
        random_state=42,
        class_weight="balanced",
        n_jobs=-1
    ))
])

rf_model.fit(X_train, y_train)

rf_results = evaluate_model(rf_model, "Random_Forest", X_test, y_test)

joblib.dump(rf_model, "/content/outputs/random_forest_model.joblib")
```

```python
num_classes = len(label_encoder.classes_)

if num_classes == 2:
    objective = "binary:logistic"
    eval_metric = "logloss"
else:
    objective = "multi:softprob"
    eval_metric = "mlogloss"

xgb_model = Pipeline(steps=[
    ("preprocessor", preprocessor),
    ("classifier", XGBClassifier(
        n_estimators=200,
        max_depth=6,
        learning_rate=0.08,
        subsample=0.85,
        colsample_bytree=0.85,
        objective=objective,
        eval_metric=eval_metric,
        random_state=42,
        n_jobs=-1
    ))
])

xgb_model.fit(X_train, y_train)

xgb_results = evaluate_model(xgb_model, "XGBoost", X_test, y_test)

joblib.dump(xgb_model, "/content/outputs/xgboost_model.joblib")
```

```python
results_df = pd.DataFrame([rf_results, xgb_results])
results_df = results_df.sort_values(by="f1_score", ascending=False)

print(results_df)

results_df.to_csv("/content/outputs/model_comparison_results.csv", index=False)
```

```python
def assign_priority(predicted_label, confidence):
    label = str(predicted_label).lower()

    if "benign" in label or "normal" in label:
        return "Likely false positive or benign"

    if confidence >= 0.85:
        return "High priority threat"
    elif confidence >= 0.65:
        return "Medium priority alert"
    elif confidence >= 0.50:
        return "Low priority alert"
    else:
        return "Manual review required"


def mitre_context(predicted_label):
    label = str(predicted_label).lower()

    if "benign" in label or "normal" in label:
        return "No active threat mapping"

    if "dos" in label or "ddos" in label:
        return "MITRE context: Impact or Denial of Service behaviour"

    if "brute" in label or "ssh" in label or "ftp" in label:
        return "MITRE context: Credential Access or brute force behaviour"

    if "web" in label or "sql" in label or "xss" in label:
        return "MITRE context: Initial Access or web application attack"

    if "bot" in label:
        return "MITRE context: Command and Control behaviour"

    if "infiltration" in label:
        return "MITRE context: Possible internal compromise or lateral movement"

    return "MITRE context: Requires analyst review"


best_model_name = results_df.iloc[0]["model"]

if best_model_name == "Random_Forest":
    best_model = rf_model
else:
    best_model = xgb_model

y_pred = best_model.predict(X_test)
y_proba = best_model.predict_proba(X_test)
confidence = y_proba.max(axis=1)

predicted_labels = label_encoder.inverse_transform(y_pred)
actual_labels = label_encoder.inverse_transform(y_test)

triage_df = pd.DataFrame({
    "actual_label": actual_labels,
    "predicted_label": predicted_labels,
    "model_confidence": confidence,
    "triage_priority": [assign_priority(label, conf) for label, conf in zip(predicted_labels, confidence)],
    "mitre_context": [mitre_context(label) for label in predicted_labels]
})

triage_df.head(20)
```

```python
triage_df.to_csv("/content/outputs/soc_triage_predictions.csv", index=False)
```

```python
import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import LSTM, Dense, Dropout
from tensorflow.keras.callbacks import EarlyStopping
from tensorflow.keras.utils import to_categorical

lstm_preprocessor = preprocessor

X_train_processed = lstm_preprocessor.fit_transform(X_train)
X_test_processed = lstm_preprocessor.transform(X_test)

if hasattr(X_train_processed, "toarray"):
    X_train_processed = X_train_processed.toarray()

if hasattr(X_test_processed, "toarray"):
    X_test_processed = X_test_processed.toarray()

X_train_lstm = X_train_processed.reshape((X_train_processed.shape[0], X_train_processed.shape[1], 1))
X_test_lstm = X_test_processed.reshape((X_test_processed.shape[0], X_test_processed.shape[1], 1))

num_classes = len(label_encoder.classes_)

if num_classes == 2:
    y_train_lstm = y_train
    y_test_lstm = y_test
    output_units = 1
    activation = "sigmoid"
    loss = "binary_crossentropy"
else:
    y_train_lstm = to_categorical(y_train, num_classes=num_classes)
    y_test_lstm = to_categorical(y_test, num_classes=num_classes)
    output_units = num_classes
    activation = "softmax"
    loss = "categorical_crossentropy"

lstm_model = Sequential([
    LSTM(64, input_shape=(X_train_lstm.shape[1], 1)),
    Dropout(0.30),
    Dense(32, activation="relu"),
    Dropout(0.20),
    Dense(output_units, activation=activation)
])

lstm_model.compile(
    optimizer="adam",
    loss=loss,
    metrics=["accuracy"]
)

early_stop = EarlyStopping(
    monitor="val_loss",
    patience=2,
    restore_best_weights=True
)

history = lstm_model.fit(
    X_train_lstm,
    y_train_lstm,
    validation_split=0.2,
    epochs=5,
    batch_size=256,
    callbacks=[early_stop],
    verbose=1
)
```

```python
if num_classes == 2:
    lstm_proba = lstm_model.predict(X_test_lstm).reshape(-1)
    lstm_pred = (lstm_proba >= 0.5).astype(int)
else:
    lstm_proba = lstm_model.predict(X_test_lstm)
    lstm_pred = np.argmax(lstm_proba, axis=1)

average_type = "binary" if num_classes == 2 else "weighted"

lstm_results = {
    "model": "LSTM",
    "accuracy": accuracy_score(y_test, lstm_pred),
    "precision": precision_score(y_test, lstm_pred, average=average_type, zero_division=0),
    "recall": recall_score(y_test, lstm_pred, average=average_type, zero_division=0),
    "f1_score": f1_score(y_test, lstm_pred, average=average_type, zero_division=0),
    "roc_auc": None
}

print(lstm_results)
print(classification_report(y_test, lstm_pred, target_names=label_encoder.classes_, zero_division=0))

cm = confusion_matrix(y_test, lstm_pred)

plt.figure(figsize=(8, 6))
plt.imshow(cm)
plt.title("Confusion Matrix - LSTM")
plt.xlabel("Predicted")
plt.ylabel("Actual")
plt.colorbar()

ticks = np.arange(len(label_encoder.classes_))
plt.xticks(ticks, label_encoder.classes_, rotation=45, ha="right")
plt.yticks(ticks, label_encoder.classes_)

for i in range(cm.shape[0]):
    for j in range(cm.shape[1]):
        plt.text(j, i, str(cm[i, j]), ha="center", va="center")

plt.tight_layout()
plt.savefig("/content/outputs/confusion_matrix_LSTM.png", dpi=300)
plt.show()

lstm_model.save("/content/outputs/lstm_model.keras")
```

```python
final_results_df = pd.DataFrame([rf_results, xgb_results, lstm_results])
final_results_df = final_results_df.sort_values(by="f1_score", ascending=False)

final_results_df.to_csv("/content/outputs/final_model_comparison_with_lstm.csv", index=False)

final_results_df
```

```python
# ============================================================
# FIX: Convert multiclass labels into binary Benign vs Attack
# This solves the "least populated class has only 1 member" error
# ============================================================

import numpy as np
import pandas as pd
from sklearn.preprocessing import LabelEncoder
from sklearn.model_selection import train_test_split

df = clean_columns(df)
TARGET_COLUMN = TARGET_COLUMN.strip().replace(" ", "_")

print("Current target column:", TARGET_COLUMN)
print("Available columns:")
print(df.columns.tolist())

if TARGET_COLUMN not in df.columns:
    raise ValueError("Target column not found. Please check the exact label column name.")

# Remove infinite values and missing target values
df = df.replace([np.inf, -np.inf], np.nan)
df = df.dropna(subset=[TARGET_COLUMN]).copy()

# Show original class distribution
print("\nOriginal class distribution:")
print(df[TARGET_COLUMN].astype(str).str.strip().value_counts())

# Convert to binary label: Benign vs Attack
def convert_to_binary_label(value):
    value = str(value).strip().lower()

    # Common benign labels across UNSW, CICIDS, NSL-KDD
    if value in ["benign", "normal", "0"]:
        return "Benign"
    else:
        return "Attack"

df["binary_label"] = df[TARGET_COLUMN].apply(convert_to_binary_label)

TARGET_COLUMN = "binary_label"

print("\nBinary class distribution:")
print(df[TARGET_COLUMN].value_counts())

# Safety check
if df[TARGET_COLUMN].nunique() < 2:
    raise ValueError(
        "Only one class found after binary conversion. "
        "This means your loaded dataset sample may contain only Benign or only Attack records. "
        "Load more rows or use another CSV/folder."
    )

# Drop columns that should not be used as features
drop_columns = [
    TARGET_COLUMN,
    "label",
    "Label",
    "attack_cat",
    "Attack_cat",
    "class",
    "Class",
    "id",
    "ID",
    "Flow_ID",
    "Timestamp",
    "timestamp",
    "Src_IP",
    "Dst_IP",
    "Source_IP",
    "Destination_IP",
    "source_file"
]

X = df.drop(columns=[col for col in drop_columns if col in df.columns], errors="ignore")
y_raw = df[TARGET_COLUMN].astype(str).str.strip()

label_encoder = LabelEncoder()
y = label_encoder.fit_transform(y_raw)

print("\nFinal features shape:", X.shape)
print("Final target classes:", label_encoder.classes_)
print("Encoded class count:")
print(pd.Series(y).value_counts())
```

```python
# ============================================================
# SAFE TRAIN TEST SPLIT
# ============================================================

class_counts = pd.Series(y).value_counts()

print("Encoded class counts before split:")
print(class_counts)

if class_counts.min() < 2:
    print("Warning: One class has fewer than 2 records. Running split without stratify.")
    stratify_option = None
else:
    stratify_option = y

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.25,
    random_state=42,
    stratify=stratify_option
)

print("Training data:", X_train.shape)
print("Testing data:", X_test.shape)

print("\nTraining class count:")
print(pd.Series(y_train).value_counts())

print("\nTesting class count:")
print(pd.Series(y_test).value_counts())
```

```python
# ============================================================
# FIXED NSL-KDD LOADER
# Handles both .txt and .arff files correctly
# ============================================================

import os
import glob
import numpy as np
import pandas as pd

nsl_folder = "/content/datasets/nsl_kdd"

print("Files inside NSL-KDD folder:")
for root, dirs, files in os.walk(nsl_folder):
    for file in files:
        print(os.path.join(root, file))


nsl_columns = [
    "duration", "protocol_type", "service", "flag", "src_bytes", "dst_bytes",
    "land", "wrong_fragment", "urgent", "hot", "num_failed_logins", "logged_in",
    "num_compromised", "root_shell", "su_attempted", "num_root", "num_file_creations",
    "num_shells", "num_access_files", "num_outbound_cmds", "is_host_login",
    "is_guest_login", "count", "srv_count", "serror_rate", "srv_serror_rate",
    "rerror_rate", "srv_rerror_rate", "same_srv_rate", "diff_srv_rate",
    "srv_diff_host_rate", "dst_host_count", "dst_host_srv_count",
    "dst_host_same_srv_rate", "dst_host_diff_srv_rate",
    "dst_host_same_src_port_rate", "dst_host_srv_diff_host_rate",
    "dst_host_serror_rate", "dst_host_srv_serror_rate",
    "dst_host_rerror_rate", "dst_host_srv_rerror_rate",
    "Label", "difficulty"
]


def load_nsl_file(file_path):
    """
    Loads NSL-KDD .txt or .arff file.
    .txt files are normal comma-separated files.
    .arff files must skip metadata lines before @data.
    """
    file_path = str(file_path)

    if file_path.lower().endswith(".arff"):
        print("Loading ARFF file:", file_path)

        data_lines = []
        start_data = False

        with open(file_path, "r", encoding="utf-8", errors="ignore") as f:
            for line in f:
                line = line.strip()

                if not line:
                    continue

                if line.lower().startswith("@data"):
                    start_data = True
                    continue

                if start_data and not line.startswith("@"):
                    data_lines.append(line)

        if len(data_lines) == 0:
            raise ValueError("No data rows found in ARFF file.")

        # Check number of columns in the first real data row
        first_row_columns = len(data_lines[0].split(","))

        temp_path = "/content/temp_nsl_arff_data.csv"
        with open(temp_path, "w", encoding="utf-8") as f:
            for line in data_lines:
                f.write(line + "\n")

        if first_row_columns == 43:
            df_temp = pd.read_csv(temp_path, names=nsl_columns)
        elif first_row_columns == 42:
            df_temp = pd.read_csv(temp_path, names=nsl_columns[:-1])
            df_temp["difficulty"] = np.nan
        else:
            raise ValueError(f"Unexpected number of columns in ARFF data: {first_row_columns}")

        return df_temp

    else:
        print("Loading TXT file:", file_path)

        # First try 43 columns
        df_temp = pd.read_csv(file_path, names=nsl_columns)

        # If it loads with wrong structure, handle it
        if df_temp.shape[1] == 42:
            df_temp = pd.read_csv(file_path, names=nsl_columns[:-1])
            df_temp["difficulty"] = np.nan

        return df_temp


# Prefer TXT files where available
all_files = glob.glob(nsl_folder + "/**/*", recursive=True)

train_file = None
test_file = None

# First preference: .txt files
for file in all_files:
    name = os.path.basename(file).lower()

    if os.path.isfile(file) and "kddtrain+" in name and name.endswith(".txt"):
        train_file = file

    if os.path.isfile(file) and "kddtest+" in name and name.endswith(".txt"):
        test_file = file

# Second preference: .arff test file if .txt is not available
if test_file is None:
    for file in all_files:
        name = os.path.basename(file).lower()

        if os.path.isfile(file) and "kddtest+" in name and name.endswith(".arff"):
            test_file = file

print("Selected train file:", train_file)
print("Selected test file:", test_file)

if train_file is None:
    raise FileNotFoundError("KDDTrain+ file was not found.")

if test_file is None:
    raise FileNotFoundError("KDDTest+ file was not found.")

train_df = load_nsl_file(train_file)
test_df = load_nsl_file(test_file)

df = pd.concat([train_df, test_df], ignore_index=True)

# Clean column names
df = clean_columns(df)

# Clean labels
df["Label"] = df["Label"].astype(str).str.strip().str.replace("'", "", regex=False)

print("NSL-KDD loaded successfully.")
print("Shape:", df.shape)
print(df.head())
print("\nOriginal label distribution:")
print(df["Label"].value_counts().head(20))
```
